# **NumPy Array Indexing and Slicing — সম্পূর্ণ নোটস**

**Indexing এবং Slicing কী?**

**Indexing** মানে array এর কোনো নির্দিষ্ট একটি উপাদান বের করা।
**Slicing** মানে array এর একটি অংশ কেটে নেওয়া।

পাইথন লিস্টের মতোই কাজ করে, কিন্তু NumPy তে 2D, 3D array তেও একইভাবে করা যায়।
**Index সবসময় 0 থেকে শুরু হয়** — মানে প্রথম উপাদানের index হলো 0।

**Slicing এর সাধারণ নিয়ম:** `arr[start:stop:step]`


**start** → কোথা থেকে শুরু (অন্তর্ভুক্ত)

**stop** → কোথায় শেষ (অন্তর্ভুক্ত নয়)

**step** → কত করে এগোবে

# **১. Vector (1D Array) এর Indexing ও Slicing**

```python
import numpy as np
```

```python
arr = np.arange(11,21)
arr
```

**Indexing — একটি নির্দিষ্ট উপাদান বের করা**

```python
arr[4]
```

**arr[4]** মানে index 4 এর উপাদান। Index 0 = 11, Index 1 = 12, Index 2 = 13, Index 3 = 14, Index 4 = 15।

**Slicing — অংশ কেটে নেওয়া**

```python
arr_slice = arr[3::2]
arr_slice
```

**arr[3::2] মানে:**

3 → index 3 থেকে শুরু (মান 14)

: → শেষ পর্যন্ত যাও

2 → 2 ধাপে এগোও

তাই ফলাফল: [14, 16, 18, 20]

এটা একটা 6 সারি, 5 কলামের matrix। দেখতে এরকম:

       col0  col1  col2  col3  col4
`row0` [  1,    2,    3,    4,    5 ]

`row1` [  6,    7,    8,    9,   10 ]

**row2** [ 11,   12,   13,   14,   15 ]

`row3` [ 16,   17,   18,   19,   20 ]

`row4` [ 21,   22,   23,   24,   25 ]

`row5` [ 26,   27,   28,   29,   30 ]

**2D Slicing — একটি অংশ কেটে নেওয়া**

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[1]
```

```python
arr[5]
```

```python
arr[0,4]
```

```python
arr[5,4]
```

```python
slice= arr[0:2,1:3]
slice
```

```python
slice_1= arr[3:6,3:5]
slice_1
```

```python
slice_2 = arr[3:,3:]
slice_2
```


**arr[3:, 3:] মানে:**


**সারি:** index 3 থেকে শেষ পর্যন্ত (row3, row4, row5)

**কলাম:** index 3 থেকে শেষ পর্যন্ত (col3, col4)

ফলাফল নিচের ডান কোণের 3×2 অংশ:

[[19, 20],

 [24, 25],

 [29, 30]]

**পুরো একটি কলাম বের করা**

```python
arr[:,2]
```

**arr[:, 2] মানে:**


: → সব সারি নাও

2 → শুধু index 2 কলাম নাও (col2)


**ফলাফল:** [ 3,  8, 13, 18, 23, 28] — পুরো তৃতীয় কলাম।

একইভাবে পুরো একটি সারি বের করতে হলে: arr[2, :] → তৃতীয় সারি

# **৩. Boolean Indexing — শর্ত দিয়ে ফিল্টার করা**

Boolean Indexing মানে কোনো শর্ত দিলে array থেকে শুধু সেই উপাদানগুলো বের হবে যেগুলো শর্ত পূরণ করে।
তিনটি ধাপে কাজ করে:

1. শর্ত দিয়ে True/False এর array তৈরি করো
2. সেই True/False array দিয়ে মূল array কে filter করো
3. শুধু True অবস্থানের মানগুলো বেরিয়ে আসে

```python
arr = np.arange(11,21)
arr
```

**ধাপ ১ — Boolean Array তৈরি**

```python
bool_index = arr % 2 == 0
bool_index
```

প্রতিটি উপাদান জোড় কিনা চেক করছে। জোড় হলে True, বিজোড় হলে False।

ফলাফল: [False, True, False, True, False, True, False, True, False, True]

**ধাপ ২ — Filter করা**

```python
arr = arr[bool_index]
arr
```

শুধু যেখানে True সেখানকার মান বের হয়।
ফলাফল: [12, 14, 16, 18, 20] — শুধু জোড় সংখ্যাগুলো।


**শর্টকাট:** এই দুটো ধাপ একলাইনেও লেখা যায়: arr[arr % 2 == 0]
