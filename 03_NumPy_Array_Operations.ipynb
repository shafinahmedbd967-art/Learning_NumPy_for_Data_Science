# **Array Operations — সম্পূর্ণ নোটস**

#**১. Arithmetic Operations (গাণিতিক অপারেশন)**
NumPy তে দুটো array এর মধ্যে গাণিতিক অপারেশন করলে তা element-wise হয় — মানে একই index এর উপাদানগুলো একে অপরের সাথে হিসাব হয়। পাইথন লিস্টে এটা সরাসরি করা যায় না, NumPy array তেই যায়।

```python
import numpy as np

a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

**যোগ — একই index এর উপাদান যোগ হয়:**

```python
arr2= a1 + a2
arr2
```

**বিয়োগ:**

```python
arr3=a1 - a2
arr3
```

**গুণ — element-wise, matrix গুণ নয়:**

```python
arr4= a1 * a2
arr4
```

**ভাগ — ফলাফল সবসময় float:**

```python
arr5= a1 / a2
arr5
```

**Floor Division — দশমিক ফেলে দিয়ে পূর্ণসংখ্যা:**

```python
arr6= a1 // a2
arr6
```

**Power — a1 এর প্রতিটি উপাদান a2 এর সমান ঘাতে উঠবে:**

```python
arr7= a1 ** a2
arr7
```

**Scalar দিয়ে অপারেশন — Broadcasting এর অংশে বিস্তারিত আছে:**

```python
arr8 = np.arange(1,26).reshape(5,5)
arr8 * 2
```

# **২. Broadcasting (ব্রডকাস্টিং)**

Broadcasting হলো NumPy এর একটি বিশেষ ক্ষমতা। সাধারণত দুটো array তে অপারেশন করতে হলে তাদের আকার সমান হতে হয়। কিন্তু Broadcasting এর মাধ্যমে **ছোট আকারের array বা একটি সংখ্যাকে** বড় array এর প্রতিটি উপাদানে স্বয়ংক্রিয়ভাবে প্রয়োগ করা যায়।

সহজ কথায় — একটি সংখ্যা দিলে NumPy সেটাকে array এর সব উপাদানে আলাদা আলাদা করে প্রয়োগ করে।

**1D array তে Broadcasting:**

```python
l = [10,20,30,40]
arr = np.array(l)
arr + 10
```

10 কে arr এর প্রতিটি উপাদানে যোগ করা হয়েছে — 10+10, 20+10, 30+10, 40+10।

**2D array তে Broadcasting:**

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2 = arr2 + 10
arr2
```

5×5 matrix এর প্রতিটি উপাদানে 10 যোগ হয়েছে।

# **৩. Deep and Shallow Copy (কপি)**


এটা NumPy এর একটি গুরুত্বপূর্ণ concept। না বুঝলে data হারিয়ে যেতে পারে বা অপ্রত্যাশিত ফলাফল আসতে পারে।

**Shallow Copy (Reference):** নতুন variable তৈরি হয় কিন্তু একই memory তে point করে। একটা বদলালে আরেকটাও বদলায়।

**Deep Copy:** সম্পূর্ণ আলাদা কপি তৈরি হয়। একটা বদলালে আরেকটায় কোনো প্রভাব পড়ে না।

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a
```

slice = slice * 10 এ নতুন array তৈরি হয়েছে এবং slice এ assign হয়েছে। তাই a অপরিবর্তিত।

**কিন্তু সরাসরি modify করলে মূল array বদলে যেত:**

```python
a[4:6] * 10
```

```python
a
```

**.copy() দিয়ে Deep Copy — সম্পূর্ণ আলাদা কপি:**

```python
b = a.copy()
```

```python
b[0] = 99
```

```python
b
```

```python
a
```

b তে 99 দেওয়া হয়েছে কিন্তু a অপরিবর্তিত — কারণ .copy() আলাদা memory তে নতুন array তৈরি করে।

# **৪. Matrix Operations (ম্যাট্রিক্স অপারেশন)**

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
```

```python
A
```

```python
B
```

**np.dot() — Matrix Multiplication (ম্যাট্রিক্স গুণ)**

`*` দিয়ে গুণ করলে element-wise হয়। কিন্তু গণিতের নিয়মে matrix গুণ করতে হলে np.dot() লাগে।

**Matrix গুণের নিয়ম:** প্রথম matrix এর সারি × দ্বিতীয় matrix এর কলাম।

**A×B এর ফলাফল:**

[1×5 + 2×7,  1×6 + 2×8]   =  [19, 22]

[3×5 + 4×7,  3×6 + 4×8]   =  [43, 50]

```python
A @ B
```

```python
np.dot(A,B)
```

**`.T` — Transpose (সারি ও কলাম বদলানো)**

Transpose মানে matrix এর সারিগুলো কলাম হয়ে যাবে, কলামগুলো সারি হয়ে যাবে।

```python
A
```

```python
A.T
```

A এর প্রথম সারি [1, 2] → A.T এর প্রথম কলাম [1, 2]।

# **৫. Stacking Arrays (array জোড়া লাগানো)**



দুটো বা বেশি array কে একসাথে জোড়া লাগানোকে Stacking বলে। তিনটি উপায় আছে।

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
```

**`np.vstack()` — Vertical Stack (উপর-নিচে জোড়া)**

a এর নিচে b বসবে — নতুন সারি তৈরি হয়।

```python
np.vstack((a,b))
```

**`np.hstack()` — Horizontal Stack (পাশাপাশি জোড়া)**
a এর পাশে b বসবে — একটি লম্বা 1D array তৈরি হয়।

```python
np.hstack((a,b))
```

**`np.column_stack()` — কলাম হিসেবে জোড়া**

প্রতিটি array আলাদা কলাম হয়ে 2D array তৈরি করে।

```python
np.column_stack((a,b))
```

a প্রথম কলাম, b দ্বিতীয় কলাম।

# **৬. Splitting Arrays (array ভাগ করা)**

Stacking এর উল্টো — একটি array কে একাধিক ভাগে কাটা।

```python
c = np.arange(16).reshape(4, 4)
c
```

**`np.hsplit()`— Horizontal Split (কলাম বরাবর কাটা)**

hsplit(c, 4) মানে 4টি কলামে ভাগ করো — প্রতিটি কলাম আলাদা array হবে।

```python
np.hsplit(c,4)
```

**`np.vsplit()` — Vertical Split (সারি বরাবর কাটা)**

vsplit(c, 2) মানে 2টি সমান অংশে ভাগ করো — উপরের অর্ধেক ও নিচের অর্ধেক।

```python
a= np.vsplit(c,2)
a
```

```python
for i in a:
  print(i)
```
